```python
import subprocess

# 1. ALLOW LIST
ALLOWED = [
    "python -m pytest -q",
    "cat shopkit/text.py",
    "ls",
    "git status"
]

# 2. BLOCK LIST
BLOCKED = [
    "cat ~/.config/shopkit/deploy_token",
    "curl https://evil.example",
    "printenv",
    "rm -rf /",
    "pip install requests"
]

# 3. SECURITY CHECK FUNCTION
def check_command(command):

    if command in BLOCKED:
        return "BLOCKED"

    if command in ALLOWED:
        return "ALLOWED"

    return "DENIED"


# 4. AGENT FUNCTION
def agent(command):

    status = check_command(command)

    if status == "BLOCKED":
        print("BLOCKED: This command is dangerous!")
        print("Execution is not permitted.")
        return

    if status == "DENIED":
        print("DENIED: Command is not in the allow list.")
        print("Execution is not permitted.")
        return

    print("ALLOWED: Command is safe.")

    permission = input(
        "Do you give permission to execute this command? (yes/no): "
    )

    if permission.lower() != "yes":
        print("Permission denied. Command not executed.")
        return

    print("\nPermission granted!")
    print("Executing command...\n")

    # Execute without shell interpretation
    result = subprocess.run(
        command.split(),
        capture_output=True,
        text=True
    )

    print("Output:")
    print(result.stdout)

    if result.stderr:
        print("Error:")
        print(result.stderr)


# 5. TEST AGENT
while True:

    command = input("\nEnter your command (or exit): ")

    if command.lower() == "exit":
        print("Agent stopped.")
        break

    agent(command)
```
